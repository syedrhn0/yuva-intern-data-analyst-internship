# Task 3: Real-World Dataset Analysis Project

Dataset: scikit-learn diabetes progression research dataset.

## Load and inspect

```
import pandas as pd
df = pd.read_csv('data/raw/diabetes_dataset.csv')
print(df.shape)
display(df.head())
df.info()
```

## Data quality

```
print(df.isna().sum())
print('Duplicates:', df.duplicated().sum())
display(df.describe().T)
```

## Exploratory analysis

```
import matplotlib.pyplot as plt
df['target'].hist(bins=25)
plt.title('Target distribution')
plt.show()
display(df.corr(numeric_only=True)['target'].sort_values(ascending=False))
```

## Model comparison

```
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
X, y = df.drop(columns='target'), df['target']
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=.2,random_state=42)
models = {'Linear Regression': LinearRegression(),
          'Ridge': make_pipeline(StandardScaler(), Ridge(alpha=1)),
          'Random Forest': RandomForestRegressor(n_estimators=200,min_samples_leaf=3,random_state=42)}
rows=[]
for name, model in models.items():
    model.fit(X_train,y_train); p=model.predict(X_test)
    rows.append({'model':name,'MAE':mean_absolute_error(y_test,p),
                 'RMSE':mean_squared_error(y_test,p)**.5,'R2':r2_score(y_test,p)})
pd.DataFrame(rows).sort_values('MAE')
```

## Interpretation and limitations

```
# Lower MAE and RMSE indicate smaller prediction errors.
# R-squared describes the share of target variation explained by the model.
# This small, pre-scaled research dataset is for learning, not clinical use.
```
